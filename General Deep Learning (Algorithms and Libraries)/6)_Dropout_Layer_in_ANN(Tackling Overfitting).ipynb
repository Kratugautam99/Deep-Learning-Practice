```python
!pip install tensorflow
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report
from tensorflow import keras as kr
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("10) numerical_sonar.csv",header=None)
df.head()
```

```python
x = df.drop(60, axis=1)
y = df[60]
y = pd.get_dummies(y, drop_first=True).astype(int)
y.sample(5)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=1)
```

```python
model = kr.Sequential([
    kr.layers.Dense(60, input_dim=60, activation='relu'),
    kr.layers.Dense(30, activation='relu'),
    kr.layers.Dense(15, activation='relu'),
    kr.layers.Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(x_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(x_test, y_test)
```

```python
model = kr.Sequential([
    kr.layers.Dense(60, input_dim=60, activation='relu'),
    kr.layers.Dropout(0.5),
    kr.layers.Dense(30, activation='relu'),
    kr.layers.Dropout(0.5),
    kr.layers.Dense(15, activation='relu'),
    kr.layers.Dropout(0.5),
    kr.layers.Dense(1, activation='sigmoid')
])

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(x_train, y_train, epochs=100, batch_size=8)
```

```python
model.evaluate(x_test, y_test)
```
