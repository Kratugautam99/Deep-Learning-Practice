```python
import tensorflow_hub as hub
import tensorflow_text as text
```

```python
preprocess_url = "https://tfhub.dev/tensorflow/bert_en_uncased_preprocess/3"
encoder_url = "https://tfhub.dev/tensorflow/bert_en_uncased_L-12_H-768_A-12/4"
```

```python
bert_preprocess_model = hub.KerasLayer(preprocess_url)
text_test = [
    'What a fantastic day it is!',
    'Coding is an art and science',
    'The weather is beautiful today',
    'Artificial intelligence is fascinating',
    'I enjoy reading books on weekends',
    'The food at that restaurant was delicious',
    'Exercise helps in staying healthy',
    'The new smartphone has amazing features',
    'Traveling to new places is exciting',
    'The concert last night was incredible'
]
text_preprocessed = bert_preprocess_model(text_test)
text_preprocessed.keys()
```

```python
print(text_preprocessed['input_mask'],"\n")
print(text_preprocessed['input_type_ids'],"\n")
print(text_preprocessed['input_word_ids'],"\n")
```

```python
bert_model = hub.KerasLayer(encoder_url)
bert_results = bert_model(text_preprocessed)
bert_results.keys()
```

```python
print(bert_results['pooled_output'],"\n")
print(bert_results['sequence_output'],"\n")
#print(bert_results['encoder_outputs'],"\n")
print(bert_results['default'],"\n")
```

```python
bert_results['encoder_outputs'][-1] == bert_results['sequence_output']
```

```python
bert_results['default'] == bert_results['pooled_output']
```
