```python
df = pd.read_csv("7) insurance_data.csv")
df.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(df[['age','affordibility']],df.bought_insurance,test_size=0.2, random_state=25)
```

```python
x_train["age"] = x_train["age"]/100
x_test["age"] = x_test["age"]/100
```

```python
KerasNeuralNetwork = keras.Sequential([
    keras.layers.Dense(1, input_shape=(2,), activation='sigmoid', kernel_initializer='ones', bias_initializer='zeros')
])

KerasNeuralNetwork.compile(optimizer=Adam(learning_rate = 0.5),
              loss='binary_crossentropy',
              metrics=['accuracy'])

KerasNeuralNetwork.fit(x_train, y_train, epochs=1000)
```

```python
KerasNeuralNetwork.evaluate(x_test,y_test)
```

```python
class KratuNN:
  def __init__(self):
    self.w1 = 1
    self.w2 = 1
    self.b = 0

  def fit(self, x_train, y_train, epochs, rate, loss_threshold):
    self.w1, self.w2, self.b = self.gradient_descent(x_train.iloc[:, 0], x_train.iloc[:, 1], y_train, epochs, rate, loss_threshold=0)

  def gradient_descent(self, w1vec, w2vec, y_true, epochs, rate, loss_threshold=0):
    w1 = w2 = 1
    b = 0
    n = len(w1vec)
    for i in range(epochs+1):
      weighted_sum = w1*w1vec + w2*w2vec + b
      y_pred = self.sigmoid(weighted_sum)
      loss = self.log_loss(y_true, y_pred)
      w1d = 1/n*np.dot(np.transpose(w1vec),(y_pred-y_true))
      w2d = 1/n*np.dot(np.transpose(w2vec),(y_pred-y_true))
      bd = np.mean(y_pred - y_true)
      w1 = w1 - rate*w1d
      w2 = w2 - rate*w2d
      b = b - rate*bd
      if i%1000==0:
        print(f"Epoch:{i}, w1:{w1}, w2:{w2}, bias:{b}, loss:{loss}")
      if loss<=loss_threshold:
        break
    print(f"Epoch:{i}, w1:{w1}, w2:{w2}, bias:{b}, loss:{loss}")
    return w1, w2, b

  def log_loss(self, y_true, y_pred):
    y_pred = [max(i, 1e-15) for i in y_pred]
    y_pred = [min(i, 1-1e-15) for i in y_pred]
    y_pred = np.array(y_pred)
    return -np.mean(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))

  def sigmoid(self, y_true):
    return 1/(1+np.exp(-y_true))

  def predict(self, x_test):
    weighted_sum = self.w1*x_test.iloc[:, 0] + self.w2*x_test.iloc[:, 1] + self.b
    return self.sigmoid(weighted_sum)

  def evaluate(self, x_test, y_test):
    y_pred_test = self.predict(x_test)
    correct = 0
    y_pred_test1 = y_pred_test.apply(lambda x: 1 if x > 0.5 else 0)
    correct = (y_pred_test1 == y_test).sum()
    return [float(self.log_loss(y_test, y_pred_test)),float(correct/len(y_pred_test))]
```

```python
KratuNeuralNetwork = KratuNN()
KratuNeuralNetwork.fit(x_train, y_train, epochs=20000, rate=0.5, loss_threshold=0.4103)
coef, intercept = KerasNeuralNetwork.get_weights()
print(f"KratuNeuralNetwork's weights/biases: w1: {KratuNeuralNetwork.w1}, w2: {KratuNeuralNetwork.w2}, b: {KratuNeuralNetwork.b}")
print(f"KerasNeuralNetwork's weights/biases: w1: {coef[0][0]}, w2: {coef[1][0]}, b: {intercept[0]}")
```
