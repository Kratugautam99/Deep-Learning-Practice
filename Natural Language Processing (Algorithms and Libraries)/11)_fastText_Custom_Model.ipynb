```python
def preprocess(text):
    text = re.sub(r'[^\w\s\']',' ', text)
    text = re.sub(r'[ \n]+', ' ', text)
    return text.strip().lower()
```

```python
df = pd.read_csv("20) indian_food.csv")
print(df.shape)
df.head(3)
df.TranslatedInstructions = df.TranslatedInstructions.map(preprocess)
df.to_csv("6) food_receipes.txt", columns=["TranslatedInstructions"], header=None, index=False)
model = fasttext.train_unsupervised("6) food_receipes.txt")
```
