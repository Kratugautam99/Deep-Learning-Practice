```python
# !pip install fasttext
# !pip install numpy==1.23.5
import pandas as pd
import re
import fasttext
from sklearn.model_selection import train_test_split
```

```python
df= pd.read_csv("21) ecommerce_data.csv", names=["category", "description"], header=None)
print(df.category.value_counts())
df.category.replace("Clothing & Accessories", "Clothing_Accessories", inplace=True)
df.dropna(inplace=True)
print(df.shape)
df.head(3)
```

```python
df["traindata"] = "__label__" + df["category"] + " " + df["description"]
def preprocess(text):
    text = re.sub(r'[^\w\s\']',' ', text)
    text = re.sub(' +', ' ', text)
    return text.strip().lower()
df["traindata"] = df["traindata"].map(preprocess)
train, test = train_test_split(df, test_size=0.2)
train.to_csv("7) ecommerce_train", columns=["traindata"], index=False, header=False)
test.to_csv("8) ecommerce_test", columns=["traindata"], index=False, header=False)
df
```

```python
model = fasttext.train_supervised(input="7) ecommerce_train")
model.test("8) ecommerce_test")
```
