```python
import spacy
nlp = spacy.blank("en")
nlp.add_pipe("sentencizer")
doc = nlp("Dr. Mario is the Surgeon of Italy. Sonic is the Olympic Champion!")
for i in doc.sents:
  print(i)
```

```python
print(type(nlp), type(doc), type(doc[0]), type(doc[0:2]), type(doc.sents), sep="\n")
```

```python
from spacy.symbols import ORTH
nlp.tokenizer.add_special_case("Mario", [{"ORTH": "Mar"}, {"ORTH": "io"}])
doc = nlp("Mario is the Surgeon of Italy. Sonic is the Olympic Champion!")
for i in doc:
  print(i)
```

```python
# Exercise 1
urls = []
text = '''
Look for data to help you address the question. Governments are good
sources because data from public research is often freely available. Good
places to start include http://www.data.gov/, and http://www.science.
gov/, and in the United Kingdom, http://data.gov.uk/.
Two of my favorite data sets are the General Social Survey at http://www3.norc.org/gss+website/,
and the European Social Survey at http://www.europeansocialsurvey.org/.
'''
doc = nlp(text)
for i in doc:
  if i.like_url:
    urls.append(i.text)
print(urls)
```
