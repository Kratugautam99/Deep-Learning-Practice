```python
import spacy
import pandas as pd
from spacy.lang.en.stop_words import STOP_WORDS
len(STOP_WORDS)
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Occult is now legal in the contemperory world.")
for token in doc:
  if token.is_stop:
    print(token)
```

```python
def removestops(text):
  doc = nlp(text)
  stop = [token.text for token in doc if not token.is_stop]
  return " ".join(stop)
removestops("Occult is now legal in the contemperory world.")
```
