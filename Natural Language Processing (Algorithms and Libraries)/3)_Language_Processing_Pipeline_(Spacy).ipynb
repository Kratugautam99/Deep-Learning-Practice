```python
import spacy
from spacy import displacy
from spacy.tokens import Span
nlp = spacy.blank("en")
doc0 = nlp("C.A. means Chartered Accountant. Ria works as C.A.!")
print(nlp.pipe_names, "| Blank NLP Pipeline")
for tok in doc0:
  print(tok)
```

```python
nlp1 = spacy.load("en_core_web_sm")
doc = nlp1("C.A. means Chartered Accountant. Ria works as C.A.!")
print(nlp1.pipe_names, "| Core NLP Pipeline\n") #nlp1.pipeline
for tok in doc:
  print(tok, "|", spacy.explain(tok.pos_), "|", spacy.explain(tok.tag_), "|", tok.lemma_)
```

```python
doc = nlp1("Tesla Inc is going to acquire Twitter Company for $45 billion")
for e in doc.ents:
  print(e, "|", e.text, "|", e.label_)
displacy.render(doc, style="ent")
```

```python
nlp.add_pipe("ner",source=nlp1)
print(nlp.pipe_names)
doc = nlp("Mukesh Ambani is a Billionaire.")
for ent in doc.ents:
    print(ent.text,"|", ent.label_)
displacy.render(doc, style="ent")
```

```python
text = '''Ravi and Raju are the best friends from school days.They wanted to go for a world tour and
visit famous cities like Paris, London, Dubai, Rome etc and also they called their another friend Mohan to take part of this world tour.
They started their journey from Hyderabad and spent next 3 months travelling all the wonderful cities in the world and cherish a happy moments!
'''
doc = nlp1(text)
prop = []
for i in doc:
  if i.pos_ == "PROPN":
    prop.append(i.text)
prop.insert(0,'Ravi')
print("Proper Nouns: ", prop, "\nCount of it: " , len(prop))
```

```python
text = '''The Top 5 companies in USA are Tesla, Walmart, Amazon, Microsoft, Google and the top 5 companies in
India are Infosys, Reliance, HDFC Bank, Hindustan Unilever and Bharti Airtel'''
doc = nlp1(text)
com = []
for ent in doc.ents:
  if ent.label_ == "ORG":
    com.append(ent.text)
print("Companies: ", com, "\nCount of it: " , len(com))
```

```python
s1 = Span(doc0, 0, 1, label='JOBTITLE')
s2 = Span(doc0, 5, 6, label='PERSON')
doc0.set_ents([s1,s2], default='unmodified')
for ent in doc0.ents:
    print(ent.text,"|", ent.label_)
```

```python
text = """Kiran wants to know the famous foods in each state of India. So, he opened Google and searched for this question.
Google showed that in Delhi it is Chaat, in Gujarat it is Dal Dhokli, in Tamilnadu it is Pongal, in Andhrapradesh it is Biryani,
in Assam it is Papaya Khar, in Bihar it is Litti Chowkha, and so on for all other states."""
doc = nlp1(text)
geo_locations = [ent.text for ent in doc.ents if ent.label_ == "GPE"]
geo_locations = list(set(geo_locations))
print("Geographical Location Names:", geo_locations)
print("Count:", len(geo_locations))
```

```python
text = """Sachin Tendulkar was born on 24 April 1973, Virat Kohli was born on 5 November 1988,
Dhoni was born on 7 July 1981, and finally Ricky Ponting was born on 19 December 1974."""
doc = nlp1(text)
birth_dates = [ent.text for ent in doc.ents if ent.label_ == "DATE"]
birth_dates = list(set(birth_dates))
print("All Birth Dates:", birth_dates)
print("Count:", len(birth_dates))
```
